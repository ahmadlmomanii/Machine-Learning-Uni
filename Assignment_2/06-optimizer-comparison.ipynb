# Optimizer Comparison

## Introduction
This notebook compares five optimization algorithms used to train Logistic Regression models for the Bank Marketing classification task.

The evaluated optimizers are:
1. Gradient Descent (GD)
2. Stochastic Gradient Descent (SGD)
3. Mini-Batch Gradient Descent
4. RMSprop
5. Adam

The comparison is based on Accuracy, Precision, Recall, F1-Score, Training Time, and Final Training Loss.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
results = pd.DataFrame([
    {
        "Optimizer": "Gradient Descent",
        "Accuracy": 0.6946,
        "Precision": 0.7132,
        "Recall": 0.5945,
        "F1-Score": 0.6485,
        "Training Time": 1.0036,
        "Final Loss": 0.5761864
    },
    {
        "Optimizer": "SGD",
        "Accuracy": 0.6932,
        "Precision": 0.7660,
        "Recall": 0.5076,
        "F1-Score": 0.6106,
        "Training Time": 12.3348,
        "Final Loss": 0.5772908
    },
    {
        "Optimizer": "Mini-Batch GD",
        "Accuracy": 0.6941,
        "Precision": 0.7198,
        "Recall": 0.5803,
        "F1-Score": 0.6426,
        "Training Time": 0.9104,
        "Final Loss": 0.5736182
    },
    {
        "Optimizer": "RMSprop",
        "Accuracy": 0.6744,
        "Precision": 0.6927,
        "Recall": 0.5624,
        "F1-Score": 0.6208,
        "Training Time": 0.1151,
        "Final Loss": 0.6270601
    },
    {
        "Optimizer": "Adam",
        "Accuracy": 0.6753,
        "Precision": 0.6970,
        "Recall": 0.5567,
        "F1-Score": 0.6190,
        "Training Time": 0.1188,
        "Final Loss": 0.6327429
    }
])

display(results.round(4))
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1-Score"]

results.set_index("Optimizer")[metrics].plot(
    kind="bar",
    figsize=(12, 6),
    width=0.8
)

plt.title("Performance Comparison of Optimization Algorithms")
plt.ylabel("Score")
plt.xlabel("Optimizer")
plt.ylim(0, 1)
plt.xticks(rotation=20)
plt.legend(loc="lower right")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(results["Optimizer"], results["Training Time"])

plt.title("Training Time Comparison")
plt.xlabel("Optimizer")
plt.ylabel("Training Time (Seconds)")
plt.xticks(rotation=20)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## Conclusion

Based on the experimental results, the five optimization algorithms showed different performance characteristics.

- **Gradient Descent** achieved the highest Accuracy (69.46%) and F1-Score (64.85%), making it the strongest optimizer in terms of overall classification performance in this experiment.

- **Stochastic Gradient Descent (SGD)** achieved the highest Precision (76.60%), but it also had the longest training time (12.33 seconds) and a lower Recall.

- **Mini-Batch Gradient Descent** achieved performance close to standard Gradient Descent, with an Accuracy of 69.41% and an F1-Score of 64.26%. It also had a relatively short training time.

- **RMSprop** had the shortest training time (0.1151 seconds), although its classification performance was lower than GD and Mini-Batch GD.

- **Adam** had a similar training time to RMSprop, but its classification performance was also lower than GD and Mini-Batch GD.

### Overall Conclusion

Gradient Descent achieved the best overall classification results in this experiment, while Mini-Batch Gradient Descent provided a good balance between performance and training time.

The results depend on the selected hyperparameters, number of epochs, and experimental environment. Therefore, they should be interpreted as results for these specific experiments rather than a universal ranking of the optimization algorithms.
